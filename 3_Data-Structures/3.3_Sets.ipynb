#### Sets
Sets are a built-in data type in Python used to store collections of unique items. They are unordered, meaning that the elements do not follow a specific order, and they do not allow duplicate elements. Sets are useful for membership tests, eliminating duplicate entries, and performing mathematical set operations like union, intersection, difference, and symmetric difference.

```python
## how to create sets
my_set={1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
my_empty_set=set([1,2,3,3,4,5,4,5,6,7,8,9,10])
print(my_empty_set)
```

```python
## Basic set operations

## Adding and removing the elements in a set

my_set.add(7)
print(my_set)

my_set.add(7)
print(my_set)
```

```python
## Removing the elements from a set

my_set.remove(3)
print(my_set)

my_set.remove(10)

```

```python
## Removing the elements from a set using discard method which enables to not throw an error if the element is not present in the set unlike remove method
my_set.discard(10)
print(my_set)
```

```python
## Pop method removes and returns an arbitrary element from the set
popped_element=my_set.pop()
print(popped_element)
print(my_set)
```

```python
## Clearing the set
my_set.clear()
print(my_set)
```

```python
## Set membership test
my_set={1,2,3,4,5}
print(3 in my_set)
print(7 in my_set)
```

```python
## Mathematical set operations
A={1,2,3,4,5}
B={4,5,6,7,8}
## Union of sets
union_set=A.union(B)
print(union_set)
## Intersection of sets
intersection_set=A.intersection(B)
print(intersection_set)

intersection_update_set=A.intersection_update(B)
print(intersection_update_set)
## Difference of sets
difference_set=A.difference(B)
print(difference_set)
## Symmetric difference of sets
symmetric_difference_set=A.symmetric_difference(B)
print(symmetric_difference_set)
```

```python
## set methods

set1={1,2,3}
set2={3,4,5}
print(set1.issubset(set2)) ## checks fi all the elements of set1 are present in set2

print(set1.issuperset(set2)) ## checks if all the elements of set2 are present in set1

print(set1.isdisjoint(set2)) ## checks if there are no common elements between set
```

```python
list1=[1,2,2,3,4,4,5]
set(list1) ## removing duplicates from the list using set
```

```python
## counting unique words in texts

text="This is a sample text with some sample words. This text is just for testing."
words=text.split()

## converting the list of words into a set in order to get the unique words from the sentence

unique_words=set(words)
print(unique_words)
print(len(unique_words))
```
