##### Introduction to Dictionaries

Dictionaries are unordered collections of items. They store data in key-value pairs.
Keys must be unique and immutable (e.g., strings, numbers, or tuples), while values can be of any type.

```python
## Creating a dictionary

empty_dict={}
print(type(empty_dict))

empt_dict=dict()
print(type(empt_dict))
```

```python
student={"name":"Kaushik","age":21,"Grade":"8.41"}
print(student)
print(type(student))
```

```python
## Error in dictionary keys
# my_dict={["name"]:"Kaushik","age":21} ## list cannot be used as a key in a dictionary
# Single key is always used instead of the last key if there are duplicate keys in a dictionary
student={"name":"Kaushik","age":21,"name":"8.41"}
print(student)
```

```python
## Accessing dictionary elements in a dictionary
student={"name":"Kaushik","age":21,"Grade":"A"}
print(student['Grade']) ## accessing value using key

## Accessing the dictionary values using the get() method
print(student.get('name')) 

print(student.get('Last_name')) ## returns None if the key is not found in the dictionary

print(student.get('Last_name','Not Found')) ## returns the default value if the key is not found in the dictionary
```

```python
## Modifying the dictionary elements
# dictionaries are mutable that is it can be modified after creation or we can add and delete the elements from the dictionary
student={"name":"Kaushik","age":21,"Grade":"A"}
print(student)

student['age']=22 ## modifying the value of an existing key in the dictionary
print(student)

student['Address']="BLR"
print(student) ## adding a new key-value pair to the dictionary

del student['Grade'] ## deleting a key-value pair from the dictionary
print(student)
```

```python
## Dictionary methods
student={"name":"Kaushik","age":21,"Grade":"A"}
print(student.keys()) ## returns a view object that contains the keys of the dictionary

values=student.values() ## returns a view object that contains the values of the dictionary
print(values)

items=student.items() ## returns a view object that contains the key-value pairs of the dictionary as tuples
print(items)
```

```python
student_copy['age']=22 ## modifying the value of an existing key in the copied dictionary
print(student_copy) ## the original dictionary is not affected by the changes made to the copied dictionary
```

```python
## iterating through a dictionary via loops over keys, values and items
student={"name":"Kaushik","age":21,"Grade":"A"}
for key in student.keys():
    print(key) ## iterating through the keys of the dictionary

for value in student.values():
    print(value) ## iterating through the values of the dictionary

for key,value in student.items():
    print(f" {key} : {value} ") ## iterating through the key-value pairs of the dictionary

```

```python
## Nested dictionaries
student1={"name":"Kaushik","age":21,"Grade":"A"}
student2={"name":"Rahul","age":22,"Grade":"B"}
students={"student1":student1,"student2":student2} ## creating a nested dictionary
print(students)
```

```python
Students={"student1":{"name":"Kaushik","age":21,"Grade":"A"},"student2":{"name":"Rahul","age":22,"Grade":"B"},"Student3":{"name":"Purab","age":20,"Grade":"C"}} ## creating a nested dictionary in a single line
print(Students)
```

```python
print(students['student1']['name']) ## accessing the value of a key in a nested dictionary
students['student2']['Grade']="A+" ## modifying the value of a key in a nested dictionary
print(students)
```

```python
## Iterating over nested dictionaries
for student_id,student_info in students.items():
    print(f"Student ID: {student_id}")
    for key,value in student_info.items():
        print(f" {key} : {value} ")
```

```python
## Dictionary comprehension

## Square of a list of numbers using dictionary comprehension
squares={x:x**2 for x in range(5)}
print(squares)
```

```python
## Conditionar dictionary comprehension
even_squares={x:x**2 for x in range(10) if x%2==0}
print(even_squares)
```
