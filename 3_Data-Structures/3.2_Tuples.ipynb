##### Introduction to Tuples
Explanation:

Tuples are ordered collections of items that are immutable.
They are similar to lists, but their immutability makes them different.


```python
## creating a tuple

empty_tuple=()
print(empty_tuple)
print(type(empty_tuple))
```

```python
lst=list()
print(lst)
```

```python
tpl=tuple()
print(type(tpl))
```

```python
numbers=tuple([1,2,3,4,5])
numbers
```

```python
mixed_tuple=(1,"hello",3.14,[1,2,3],(4,5))
print(mixed_tuple)
```

```python
## Accessing tuple elements
numbers[0]
```

```python
print(numbers[-1])
numbers[1:4]
```

```python
numbers[::-1]
```

```python
## Tuple Operations

concat_tuple=numbers + mixed_tuple
print(concat_tuple)
```

```python
mixed_tuple*2
```

```python
numbers*3
```

```python
##Immutable nature of tuples
## Tuples are immutable, which means you cannot change their elements after they are created.
# This will raise an error
# numbers[0] = 10

lst=[1,2,3,4,5,6]
print(lst)
lst[1]="Kaushik"
print(lst)

numbers[1]=20 ## This will raise an error because tuples are immutable
print(numbers)
```

```python
## Tuple methods
tpl=(1,2,3,4,5,2,2)
print(tpl.count(2)) ## Count the number of occurrences of 2 in the tuple
print(tpl.index(3)) ## Find the index of the first occurrence of 3 in the tuple

```

```python
## packing and unpacking tuples
## Packing a tuple means creating a tuple by assigning multiple values to a single variable. Unpacking a tuple means extracting the values from a tuple and assigning them to individual variables.
packed_tuple=1,"hello",3.14
print(packed_tuple)
```

```python
## Unpacking the tuple
a,b,c=packed_tuple
print(a)
print(b)
print(c)
```

```python
## unpacking with star operator
numbers=(1,2,3,4,5)
a,*b,c=numbers
print(a)
print(b)
print(c)
```

```python
## examples of nested tuples
lst1=((1,2,7,9,8,6),(3,5,6,4),(1,"hello",3.14,"c"))
lst1[0][2]
print(lst1[0][2])
print(lst1[2][1])
```

```python
## Iterating through nested tuple
for sub_tuple in lst1:
    for item in sub_tuple:
        print(item,end=" ")
    print()
```
