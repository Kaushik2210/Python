#### Real-world applications of lists in python

```python
## Ex:01   Manage a to do list to keep track of the tasks you need to complete. You can add tasks, mark them as completed, and view the list of pending tasks.

to_do_list=["Buy groceries","Clean the house","Pay bills"] 

## Adding to tasks
to_do_list.append("Go for a walk") ## adding a task to the to do list
to_do_list.append("Read a book") ## adding another task to the to do list

## Removing the completed tasks
to_do_list.remove("Clean the house") ## removing a task from the to do list

## Checking if a task is in the list
if "Pay bills" in to_do_list:
    print("You need to pay bills") ## checking if a task is in the to do list and printing a message if it is
    
## Viewing the pending tasks
print("Pending tasks:")
for task in to_do_list:
    print(f"- {task}") ## printing the pending tasks in the to do list

```
