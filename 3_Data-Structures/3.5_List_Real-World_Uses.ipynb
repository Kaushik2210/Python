#### Real-world applications of lists in python

```python
## Ex:01   Manage a to do list to keep track of the tasks you need to complete. You can add tasks, mark them as completed, and view the list of pending tasks.

to_do_list=["Buy groceries","Clean the house","Pay bills"] 

## Adding to tasks
to_do_list.append("Go for a walk") ## adding a task to the to do list
to_do_list.append("Read a book") ## adding another task to the to do list

## Removing the completed tasks
to_do_list.remove("Clean the house") ## removing a task from the to do list

## Checking if a task is in the list
if "Pay bills" in to_do_list:
    print("You need to pay bills") ## checking if a task is in the to do list and printing a message if it is
    
## Viewing the pending tasks
print("Pending tasks:")
for task in to_do_list:
    print(f"- {task}") ## printing the pending tasks in the to do list

```

```python
## Ex:02 Organising student grades
## Create a list to store and calculate thr average gradesfor students

student_grades=[85,90,78,92] ## creating a list of student grades
student_grades.append(88) ## adding a grade to the list
student_grades.append(95) ## adding another grade to the list
average_grade=sum(student_grades)/len(student_grades) ## calculating the average grade
print(f"The average grade is: {average_grade}") ## printing the average grade
```

```python
## Ex-03 Managing an inventory system for a store. You can add items to the inventory, update their quantities, and check if an item is in stock.
## use a list to manage inventoru items in a store

inventory=["Apples","Bananas","Oranges"] ## creating a list of inventory items
inventory.append("Grapes") ## adding an item to the inventory
inventory.append("Mangoes") ## adding another item to the inventory
if "Bananas" in inventory:
    print("Bananas are in stock") ## checking if an item is in the inventory and printing a message if it is
if "Grapes" in inventory:
    print("Grapes are in stock") ## checking if another item is in the inventory and printing a message if it is
inventory.remove("Oranges") ## removing an item from the inventory
print("Current inventory:")
for item in inventory:
    print(f"- {item}") ## printing the current inventory items

```

```python
## Ex:04
##Colleting user feedback and storing it in a list for analysis. You can add feedback entries, view all feedback, and calculate the average rating.
feedback_list=[] ## creating an empty list to store user feedback
feedback_list.append({"user":"Alice","rating":4,"comment":"Great product!"}) ## adding a feedback entry to the list
feedback_list.append({"user":"Bob","rating":5,"comment":"Excellent service!"}) ##
    ##adding another feedback entry to the list
feedback_list.append({"user":"Charlie","rating":3,"comment":"Good, but could be better."}) ## adding another feedback entry to the list
total_rating=0 ## initializing a variable to calculate the total rating
for feedback in feedback_list:
    total_rating+=feedback["rating"] ## adding the rating from each feedback entry to the total rating
average_rating=total_rating/len(feedback_list) ## calculating the average rating
print(f"The average rating is: {average_rating}") ## printing the average rating
print("User feedback:")
for feedback in feedback_list:
    print(f"{feedback['user']} rated {feedback['rating']} and commented: {feedback['comment']}") ## printing each feedback entry with user, rating, and comment
```
