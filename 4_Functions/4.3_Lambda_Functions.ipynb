##### Lambda Functions in Python
Lambda functions are small anonymous functions defined using the **lambda** keyword. They can have any number of arguments but only one expression. They are commonly used for short operations or as arguments to higher-order functions.



```python
## Syntax
lambda arguments: expression
```

```python
def add(a,b):
    return a+b
# Example usage
result = add(5, 3)
print(f"The sum is: {result}")
```

```python
addition=lambda a,b: a+b
print(type(addition))
result=addition(5,3)
print(f"The sum is: {result}")
```

```python
def even(x):
    if x%2==0:
        return True

print(even(4))
```

```python
even= lambda x: x%2==0
print(even(4))
```

```python
def add(x, y,z):
    return x + y + z

result = add(1, 2, 3)
print(f"The sum is: {result}")
```

```python
#using lambda function with multiple arguments
add= lambda x,y,z: x+y+z
result=add(1,2,3)
print(f"The sum is: {result}")
```

```python
## map function with lambda function
## map function applies a function to all items in a list (or any iterable) and returns a map object (which is an iterator).
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x**2, numbers))
print(squared)
```
