```python
## Why functions?
num=24
if num%2==0:
    print("Even")
else:
    print("Odd")
```

```python
def check_even_odd(num):
    """Check if a number is even or odd."""
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"

## Call this function with a number to check if it's even or odd
result = check_even_odd(24)
print(result)
```

```python
## Functions with multiple parameters
def add_numbers(a, b):
    """Return the sum of two numbers."""
    return a + b

sum_result = add_numbers(5, 7)
print(sum_result)
```

```python
## Default parameter values
def greet(name="Guest"):
    """Greet a person with their name."""
    return f"Hello, {name}!"

print(greet())  # Uses default value
print(greet("Alice"))  # Overrides default value
```

```python
## Variable length arguments

## Positional and keyword arguments

def print_numbers(*args):
    """Print all numbers passed as arguments."""
    for num in args:
        print(num)
print_numbers(1, 2, 3, 4, 5)

## Keywords arguments
def print_info(**kwargs):
    """Print key-value pairs passed as keyword arguments."""
    for key, value in kwargs.items():
        print(f"{key}: {value}")
print_info(name="Alice", age=30, city="New York")
```

```python
## positional and keyword arguments together
def print_info(*args,**kwargs):
    for num in args:
        print(num)  
    
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_info(1, 2, 3, name="Alice", age=30, city="New York")
```

```python
## Return statements
def multiply(a, b):
    """Return the product of two numbers."""
    return a * b

result = multiply(5, 7)
print(result)
```

```python
## Returning multiple values from a function
def calculate(a, b):
    """Return the sum and product of two numbers."""
    sum_result = a + b
    product_result = a * b
    return sum_result, product_result
sum_val, product_val = calculate(5, 7)
print(f"Sum: {sum_val}, Product: {product_val}")
```
