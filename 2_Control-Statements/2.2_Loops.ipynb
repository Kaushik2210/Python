```python
## while loop
## while loop is used to execute a block of code repeatedly until a certain condition is met.
i=0
while i<5:
    print(i)
    i+=1
```

```python
count=0
while count%2==0:
    print(count)
    count+=1
```

```python
## Loop control statements

## break statement is used to exit a loop when a certain condition is met.
for i in range(10):
    if i==5:
        break
    print(i)
```

```python
## Continue statement is used to skip the current iteration of a loop when a certain condition is met.
for i in range(10):
    if i%2==0:
        continue
    print(i)
```

```python
## pass 
## pass statement is a nulll operation, it does nothing.

for i in range(5):
    if i==3:
        print("i is 3, skipping...")
        pass
    print(i)
```

```python
## Nested Loops 
for i in range(3):
    for j in range(2):
        print(f"i: {i}, j: {j}")
```

```python
## Calculate the sum of first n natural numbers using  while loop and for loop
n=5

# using for loop
sum=0
for i in range(1,n+1):
    sum+=i
print(f"Sum of first {n} natural numbers using for loop: {sum}")

# using while loop
sum=0
i=1
while i<=n:
    sum+=i
    i+=1
print(f"Sum of first {n} natural numbers using while loop: {sum}")
```

```python
## Display prime numbers between 1 and 100
for num in range(1,101):
    if num>1:
        for i in range(2,num):
            if num%i==0:
                break
        else:
            print(num)
```
